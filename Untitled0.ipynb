```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("funnel_events_sample.csv")
df.head(10)
```

```python
#covert time
df["timestamp"] = pd.to_datetime(df["timestamp"])
```

```python
df = df.drop_duplicates(subset=["user_id", "step"])

```

```python
funnel_steps = [
    "visited_site",
    "signup_started",
    "details_filled",
    "email_verified",
    "purchase_completed"
]
```

```python
stage_counts = (
    df.groupby("step")["user_id"]
      .nunique()
      .reindex(funnel_steps)
)
```

```python
results = pd.DataFrame({
    "Stage": stage_counts.index,
    "Users": stage_counts.values
})
```

```python
conversion = [None]
users_lost = [None]
for i in range(1, len(results)):
    previous = results.loc[i-1, "Users"]
    current = results.loc[i, "Users"]

    conversion.append(round((current / previous) * 100, 2))
    users_lost.append(previous - current)

results["Conversion Rate (%)"] = conversion
results["Users Lost"] = users_lost
```

```python

drop_index = results.loc[1:, "Conversion Rate (%)"].idxmin()

print("Funnel Analysis\n")
print(results)

print("\nBiggest Drop-off")
print("----------------")

print(f"{results.loc[drop_index-1, 'Stage']} -> {results.loc[drop_index, 'Stage']}")
print(f"Users Lost: {results.loc[drop_index, 'Users Lost']}")
print(f"Conversion Rate: {results.loc[drop_index, 'Conversion Rate (%)']}%")
```

```python
drop_index = results.loc[1:, "Conversion Rate (%)"].idxmin()

print("🚩 Biggest Drop-off (auto-detected, by rate) ***")
print("----------------")
print(f"{results.loc[drop_index-1, 'Stage']} -> {results.loc[drop_index, 'Stage']}")
print(f"Conversion Rate: {results.loc[drop_index, 'Conversion Rate (%)']}%")
print(f"Users Lost: {results.loc[drop_index, 'Users Lost']}")
print("⚠ ALERT: Biggest Drop-off (auto-detected, by rate)")
```

```python
# Plot chart
plt.figure(figsize=(8,5))
plt.bar(results["Stage"], results["Users"], color="gold")

plt.title("Funnel Analysis")
plt.xlabel("Stage")
plt.ylabel("Unique Users")

plt.xticks(rotation=20)

plt.show()

```

```python
colors = ["skyblue", "salmon", "lightgreen", "gold", "orchid"]


plt.figure(figsize=(7, 7))
plt.pie(
    results["Users"],
    labels=results["Stage"],
    autopct="%1.1f%%",
    startangle=90,
    colors=colors,
)
plt.title("Funnel — Share of Users by Stage")
plt.tight_layout()
plt.savefig("funnel_pie_chart.png", dpi=150)
plt.show()
```

```python

#  Recommendation

print("\nRecommendation:")
print(
    f"The largest drop-off occurs between "
    f"{results.loc[drop_index-1, 'Stage']} and "
    f"{results.loc[drop_index, 'Stage']}. "
    "This stage should be investigated to improve user conversion."
)
```

```python
# Time-to-convert — avg minutes between two consecutive stages
pivot = df.pivot_table(index="user_id", columns="step", values="timestamp", aggfunc="first")
pivot = pivot[funnel_steps]

print("\nTime-to-Convert")
print("----------------")
for i in range(1, len(funnel_steps)):
    stage_from = funnel_steps[i-1]
    stage_to = funnel_steps[i]

    valid_users = pivot.dropna(subset=[stage_from, stage_to])

    if len(valid_users) == 0:
        print(f"{stage_from} -> {stage_to}: no users have both timestamps logged "
              f"(data gap in this sample, can't be calculated)")
        continue

    minutes = (valid_users[stage_to] - valid_users[stage_from]).dt.total_seconds() / 60
    print(f"{stage_from} -> {stage_to}: avg {minutes.mean():.1f} min (n={len(valid_users)} users)")
```

```python

# Segment comparison
df["id_number"] = df["user_id"].str.replace("U", "").astype(int)
df["segment"] = df["id_number"].apply(lambda x: "Even ID" if x % 2 == 0 else "Odd ID")

print("\nSegment Comparison")
print("----------------")
for segment in ["Even ID", "Odd ID"]:
    segment_data = df[df["segment"] == segment]
    seg_counts = segment_data.groupby("step")["user_id"].nunique().reindex(funnel_steps)
    print(f"\n{segment}:")
    print(seg_counts)
```

```python
# ------------------------
# Recommendation
# ------------------------
stage_before = results.loc[drop_index-1, "Stage"]
stage_after = results.loc[drop_index, "Stage"]
rate = results.loc[drop_index, "Conversion Rate (%)"]

print(f"""
Recommendation
--------------
The biggest drop happens between {stage_before} and {stage_after}.
Only {rate}% of users who reached {stage_before} actually made it to {stage_after}.
This is the stage where we lose the highest percentage of users, even though
an earlier stage loses more people in total count.

A likely reason is that users are not verifying their email in time —
maybe the email is going to spam, arriving late, or the verification step
is confusing or feels unnecessary.

Suggested fix: add a 'Resend verification email' button on the page,
and send a reminder (SMS or push notification) if the user hasn't
verified within 10 minutes. This directly targets the stage where we
are losing the most users.
""")
```
